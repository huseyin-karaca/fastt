Every table of the paper, from the records the other notebooks wrote. Each table is shown below, and written as the
manuscript inputs it — the float with its caption and label — to `tables/<name>.tex`. Nothing is computed here that
a notebook has not recorded. Run against the published records, every file is identical to the one in the
repository's `tables/`, so `git diff tables/` after a run is the check.

The records are read from the Hugging Face dataset; to use records you wrote yourself (a level-2 `main_<corpus>`
writes `main_<corpus>.json` next to the notebook), set `FASTT_RECORDS_DIR` to their folder before the import.
